# Do marginal statistics replace Scat4? Single- and cross-scale statistics
Reads the outputs of `turbulence/cross_scale_compare.py` (run it first, see its docstring). Runs: M3 (unconstrained windows), A (constrained windows), B (A + all window masses), C (B + up to 6 windows, a second tail cut at q.999 and the marginals of the imaginary parts), and the run with Scat4 (`mass_final`), seed 900, vs the data.

**How to read the bands.** Grey = data ± 2σ, where σ combines the data's own sampling noise (random halves of the data) and the run-to-run noise (spread over the 5 nested-M3 seeds, assumed the same for A, B and the Scat4 run). A run inside the band reproduces that statistic as well as one can tell with this many signals.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display
OUT = Path('figures/cross_scale_M3')   # relative to turbulence/
assert (OUT / 'summary.csv').exists(), 'run turbulence/cross_scale_compare.py first'
```

## Summary: fraction of points within 2σ of the data, per statistic family
1 = recovered everywhere. Compare each no-Scat4 run with the Scat4 run in the last column.

```python
S = pd.read_csv(OUT / 'summary.csv')
display(S.pivot(index='family', columns='run', values='frac_within_2sigma').round(2))
display(Image(OUT / 'summary.png'))
```

## Cross-scale statistics
C_pq(τ*, τ) = ⟨|δ_τ* x|^p |δ_τ x|^q⟩ / (S_p(τ*) S_q(τ)): how much large increments at one scale come together with large increments at another, at the same time. (2,2) = paper's C4, (3,3) = C6, (4,2) = C42, plus (4,4) and (3,1). First the paper curves and 1-D cuts, then the full maps as log(run/data) (white = match).

```python
display(Image(OUT / 'cross_scale_cuts.png'))
display(Image(OUT / 'cross_scale_maps.png'))
```

## Bursts in time and extremes per signal
Burst persistence: correlation between the log-size of an increment now and D steps later (a cascade keeps it high over long D). Extremes: quantiles of the largest increment of each signal, in units of the data's σ_τ.

```python
display(Image(OUT / 'persistence_extremes.png'))
```

## Scat4 statistics and wavelet bands
Scat4 panels: (run − data)/σ for every Scat4 statistic; outside ±2 = not reproduced. The Scat4 run matches them by construction, so this shows directly what the runs without Scat4 miss.

```python
display(Image(OUT / 'wavelet_scat4.png'))
```
